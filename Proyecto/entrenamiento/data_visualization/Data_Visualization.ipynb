```python
import pandas as pd
from src.Utils import plot_histograms
from sklearn.preprocessing import MinMaxScaler
```

```python
X = pd.read_csv('../dataset/dataset_in.txt', index_col=0)
Y = pd.read_csv('../dataset/dataset_out.txt', index_col=0)
```

```python
x = pd.read_csv('../dataset/dataset_in.txt', index_col=0).drop(columns=['P2', 'Q2'])
y_columns = [f'V{i}' for i in range(2,15)] + ['PERD', 'QGEN7'] #V1 ES SIEMPRE 1.0 EN DATASET
y = pd.read_csv('../dataset/dataset_out.txt', index_col=0).loc[:, y_columns]
#FILTRAR DATOS ERRÓNEOS (PÉRDIDAS NEGATIVAS)
mask = y['PERD'] >= 0
x_clean = x[mask]
y_clean = y[mask]
normalizer_x = MinMaxScaler(feature_range=(0, 1))
x_clean = pd.DataFrame(normalizer_x.fit_transform(x_clean), columns=x_clean.columns, index=x_clean.index)
```

```python
plot_histograms(X, 'Distribución de Atributos de Entrada', '../figures/data/hist_data_input.png')
plot_histograms(Y, 'Distribución de Atributos de Salida', '../figures/data/hist_data_output.png')
```

```python
plot_histograms(x_clean, 'Distribución de Atributos de Entrada', '../figures/data/hist_clean_data_input.png')
plot_histograms(y_clean, 'Distribución de Atributos de Salida', '../figures/data/hist_clean_data_output.png')
```
