```python
import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score, mean_squared_error
from sklearn.model_selection import train_test_split
```

```python
data = pd.read_csv('dataset.csv', index_col=0)
data.head()
```

**TRATAMIENTO INICIAL DATOS**

```python
np.random.seed(357823)
```

```python
attributes = data.iloc[:, :-16].drop(columns=['P2', 'Q2'])
objectives = data.iloc[:, 28:]
```

```python
(training_attributes, test_attributes, training_objetive, test_objective) = train_test_split(attributes, objectives, test_size=0.2, random_state=42)
```

**ENTRENAMIENTO MODELOS INICIALES**

MODELO 1

```python
model1 = RandomForestRegressor(n_estimators=50, max_depth=15, min_samples_split=5)
model1.fit(training_attributes, training_objetive)
```

```python
predictions = model1.predict(test_attributes)
print('MSE: ' + str(mean_squared_error(test_objective, predictions)))
print('R^2: ' + str(r2_score(test_objective, predictions)))
pd.DataFrame([mean_squared_error(test_objective, predictions, multioutput='raw_values')], columns=test_objective.columns)
```

MODELO 2

```python
attributes1 = data.iloc[:, :-16].drop(columns=['P2', 'Q2'])
objectives1 = data.iloc[:, -1:]
attributes2 = pd.concat([attributes1, objectives1], axis=1)
objectives2 = data.iloc[:, 28:-1]
```

```python
(training_attributes1, test_attributes1, training_objetive1, test_objective1) = train_test_split(attributes1, objectives1, test_size=0.2, random_state=42)
(training_attributes2, test_attributes2, training_objetive2, test_objective2) = train_test_split(attributes2, objectives2, test_size=0.2, random_state=42)
```

```python
model2_1 = RandomForestRegressor(n_estimators=50, max_depth=15, min_samples_split=5)
model2_1.fit(training_attributes1, training_objetive1.values.ravel())
model2_2 = RandomForestRegressor(n_estimators=50, max_depth=15, min_samples_split=5)
model2_2.fit(training_attributes2, training_objetive2)
```

```python
predictions1 = model2_1.predict(test_attributes1)
print('MSE: ' + str(mean_squared_error(test_objective1, predictions1)))
print('R^2: ' + str(r2_score(test_objective1, predictions1)))
pd.DataFrame([mean_squared_error(test_objective1, predictions1, multioutput='raw_values')], columns=test_objective1.columns)
```

```python
predictions2 = model2_2.predict(test_attributes2)
print('MSE: ' + str(mean_squared_error(test_objective2, predictions2)))
print('R^2: ' + str(r2_score(test_objective2, predictions2)))
pd.DataFrame([mean_squared_error(test_objective2, predictions2, multioutput='raw_values')], columns=test_objective2.columns)
```
