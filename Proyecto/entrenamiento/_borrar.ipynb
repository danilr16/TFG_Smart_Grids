```python
import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import GridSearchCV, cross_validate
from sklearn.preprocessing import MinMaxScaler
from xgboost import XGBRegressor
```

```python
in_data = pd.read_csv('dataset/dataset_in.txt', index_col=0)
out_dat = pd.read_csv('dataset/dataset_out.txt', index_col=0)
data = pd.concat([in_data, out_dat], axis=1)
data.head()
```

**TRATAMIENTO INICIAL DATOS**

```python
np.random.seed(357823)
```

```python
attributes = data.iloc[:, :-28].drop(columns=['P2', 'Q2'])
objectives = data.iloc[:, 28:].drop(columns=['I0_1', 'I0_12', 'I1_2', 'I2_3', 'I3_4', 'I4_5', 'I5_6', 'I7_8', 'I8_3', 'I8_9', 'I9_10', 'I10_11', 'I12_13']) #TODO: INCLUIR CORRIENTE Y TRATAR NAN
```

```python
normalizer = MinMaxScaler(feature_range=(0, 1))
attributes = pd.DataFrame(normalizer.fit_transform(attributes), columns=attributes.columns)
```

**ENTRENAMIENTO MODELOS INICIALES**

MODELO 1

```python
model1 = RandomForestRegressor(n_estimators=50, max_depth=15, min_samples_split=5)
results1 = cross_validate(model1, attributes, objectives, scoring=['neg_mean_absolute_error', 'r2'], cv=10, n_jobs=-1)
print('MSE: ' + str(-results1['test_neg_mean_absolute_error'].mean()))
print('R2: ' + str(results1['test_r2'].mean()))
```

MODELO 2

```python
attributes1 = data.iloc[:, :-28].drop(columns=['P2', 'Q2'])
objectives1 = data.iloc[:, -1:]
attributes2 = pd.concat([attributes1, objectives1], axis=1)
objectives2 = data.iloc[:, 28:-1].drop(columns=['I0_1', 'I0_12', 'I1_2', 'I2_3', 'I3_4', 'I4_5', 'I5_6', 'I7_8', 'I8_3', 'I8_9', 'I9_10', 'I10_11', 'I12_13'])
```

```python
model2_1 = RandomForestRegressor(n_estimators=50, max_depth=15, min_samples_split=5)
results2_1 = cross_validate(model2_1, attributes1, objectives1, scoring=['neg_mean_absolute_error', 'r2'], cv=10, n_jobs=-1, return_estimator=True)
print('MSE: ' + str(-results2_1['test_neg_mean_absolute_error'].mean()))
print('R2: ' + str(results2_1['test_r2'].mean()))
model2_2 = RandomForestRegressor(n_estimators=50, max_depth=15, min_samples_split=5)
results2_2 = cross_validate(model2_2, attributes2, objectives2, scoring=['neg_mean_absolute_error', 'r2'], cv=10, n_jobs=-1, return_estimator=True)
print('MSE: ' + str(-results2_2['test_neg_mean_absolute_error'].mean()))
print('R2: ' + str(results2_2['test_r2'].mean()))
```

```python
model2_1_best = results2_1['estimator'][2]
model2_2_best = results2_2['estimator'][6]
```

**Modelos con XGBoost**

```python
modelXGB = XGBRegressor(n_estimators=50, max_depth=15, min_samples_split=5, learning_rate=0.1, n_jobs=1, booster='gbtree')
resultsXGB = cross_validate(modelXGB, attributes, objectives, scoring=['neg_mean_absolute_error', 'r2'],
                            cv=10, n_jobs=-1, return_estimator=True, return_train_score=True)
print('MSE: ' + str(-resultsXGB['test_neg_mean_absolute_error'].mean()))
print('R2: ' + str(resultsXGB['test_r2'].mean()))
df_results = pd.DataFrame(resultsXGB)
df_results
```

```python
modelXGB2_1 = XGBRegressor(n_estimators=50, max_depth=15, min_samples_split=5, learning_rate=0.1, n_jobs=1, booster='gbtree')
resultsXGB2_1 = cross_validate(modelXGB2_1, attributes1, objectives1, scoring=['neg_mean_absolute_error', 'r2'], cv=10, n_jobs=-1, return_estimator=True, return_train_score=True)
print('MSE: ' + str(-resultsXGB2_1['test_neg_mean_absolute_error'].mean()))
print('R2: ' + str(resultsXGB2_1['test_r2'].mean()))
modelXGB2_2 = XGBRegressor(n_estimators=50, max_depth=15, min_samples_split=5, learning_rate=0.1, n_jobs=1, booster='gbtree')
resultsXGB2_2 = cross_validate(modelXGB2_2, attributes2, objectives2, scoring=['neg_mean_absolute_error', 'r2'], cv=10, n_jobs=-1, return_estimator=True, return_train_score=True)
print('MSE: ' + str(-resultsXGB2_2['test_neg_mean_absolute_error'].mean()))
print('R2: ' + str(resultsXGB2_2['test_r2'].mean()))
```
